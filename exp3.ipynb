```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


gray = cv2.imread("parrot.png", cv2.IMREAD_GRAYSCALE)

if gray is None:
    print("Error: parrot.png not found.")
    exit()

gray_eq = cv2.equalizeHist(gray)

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.hist(gray.ravel(), 256, [0, 256], color='black')
plt.title("Original Histogram")

plt.subplot(2, 2, 3)
plt.imshow(gray_eq, cmap='gray')
plt.title("Equalized Grayscale Image")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.hist(gray_eq.ravel(), 256, [0, 256], color='black')
plt.title("Equalized Histogram")

plt.tight_layout()
plt.show()

img = cv2.imread("parrot.png")

if img is None:
    print("Error: parrot.png not found.")
    exit()

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Color Image")
plt.axis("off")

plt.subplot(2, 2, 2)

colors = ('b', 'g', 'r')
for i, col in enumerate(colors):
    hist = cv2.calcHist([img], [i], None, [256], [0, 256])
    plt.plot(hist, color=col)

plt.title("Original RGB Histogram")
plt.xlim([0, 256])

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

h, s, v = cv2.split(hsv)
v_eq = cv2.equalizeHist(v)

hsv_eq = cv2.merge((h, s, v_eq))

result = cv2.cvtColor(hsv_eq, cv2.COLOR_HSV2BGR)

result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)

plt.subplot(2, 2, 3)
plt.imshow(result_rgb)
plt.title("Equalized Color Image")
plt.axis("off")

plt.subplot(2, 2, 4)

for i, col in enumerate(colors):
    hist = cv2.calcHist([result], [i], None, [256], [0, 256])
    plt.plot(hist, color=col)

plt.title("Equalized RGB Histogram")
plt.xlim([0, 256])

plt.tight_layout()
plt.show()
```
